ME CREO UNA COPIA Y ELIMINO LAS LINEAS

```python
bmw2 = bmw.copy()
```

```python
bmw2.dropna(subset=["modelo"], inplace=True)
bmw2.dropna(subset=["km"], inplace=True)
bmw2.dropna(subset=["potencia"], inplace=True)
bmw2.dropna(subset=["tipo_gasolina"], inplace=True)
bmw2.dropna(subset=["volante_regulable"], inplace=True)
bmw2.dropna(subset=["camara_trasera"], inplace=True)
bmw2.dropna(subset=["elevalunas_electrico"], inplace=True)
bmw2.dropna(subset=["precio"], inplace=True)
bmw2.dropna(subset=["fecha_venta"], inplace=True)
```

```python
bmw2.drop(columns="marca", inplace=True) #ELIMINO TAMBIEN LA COLUMNA MARCA QUE NO ERA RELEVANTE
```

```python
bmw2.isnull().sum()
```

```python
bmw2.reset_index(drop=True, inplace=True) #RESETEO INDICE
```
